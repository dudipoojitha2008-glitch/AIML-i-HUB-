```python
#Strings
#five methods of strings
my_str='peak summer'
print("Capitalised string:",my_str.capitalize())
print("Title:",my_str.title())
print("Does my string start with peak?",my_str.startswith('peak'))
print("what is the index of summer?,A:",my_str.find('summer'))
print("How many times is the letter m repeated?,A:",my_str.count('e'))
```

```python
#Lists
#five methods of Lists
my_List=[23,'Cloud',2.2,'Sky','blue']
print(my_List)
my_List.append('spread')
print(my_List)
my_List.pop(0)
print(my_List)
my_List.remove(2.2)
print(my_List)
my_List.reverse()
print(my_List)
index1=my_List.index('Sky')
print(index1)
my_List.clear()
```

```python
#Tuples
#five methods of tuples
my_tuple=((2,3), 'thunder','Sky',2.29,'Dark')
print(my_tuple)
print(my_tuple.count((2,3)))
position=my_tuple.index('Sky')
print(position)
number=(23,49,18,7,2.9)
max_ele=max(number)
print(max_ele)
sorted_numbers=sorted(number)
print(sorted_numbers)
print(len(my_tuple))
```

```python
#Sets
#five methods of sets
my_set={1,2,3}
print(my_set)
my_set.add(4)
print(my_set)
set_A={1,2,3,4,5}
set_B={6,7,8,9,10}
print(set_A.union(set_B))
print(set_A.issubset(set_B))
my_set.update(set_B)
print(my_set)
my_set.intersection_update(set_A)
print(my_set)
```

```python
#Dictionaries
#five methods of dictionary
my_dict={"name":"Poojitha","class":"b-tech 2","branch":"CSE"}
print(my_dict)
my_dict.update({"college":"BVRIT"})
print(my_dict)
removed_details=my_dict.pop("class")
print(removed_details)
print(my_dict)
last_item=my_dict.popitem()
print(last_item)
print(my_dict)
dup=my_dict.copy()
print(dup)
print(dup.get("class"))
```

```python
import numpy as np
r1=np.array([1,1,2,2,3])
r2=np.array([2,2,3,3,4])
r3=np.array([3,3,4,4,5])
r4=np.array([4,4,5,5,6])
r5=np.array([5,5,6,6,7])
matrix=np.vstack((r1, r2, r3, r4, r5))
print(matrix)
```

```python
A=np.array([[1,2,3,1,2],
            [4,5,6,4,5],
            [7,8,9,7,8],
            [1,2,3,1,2],
            [4,5,6,4,5]])
dot_product=np.dot(A,matrix)
print(dot_product)
```
